```python
#Proyek Akhir

import pandas as pd

#Data Loading
df = pd.read_csv('data_kantin.csv')

#Data Inspection
print("===HEAD===")
print(df.head())

print("===INFO===")
print(df.info())

print("===DESCRIBE===")
print(df.describe())

print("/n===SHAPE===")
print(df.shape)

#Data Cleaning
print("\n===MISSING VALUE===")
print(df.isnull().sum())

df['terjual'] = df['terjual'].fillna(0)

df = df.dropna(subset=['menu'])
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int)

print("\njumlah data setelah cleaning:", df.shape)

#Filtering
laris = df[df['terjual'] > 20]

#Sorting
urut = df.sort_values(by='terjual', ascending=False)

#Kolom turunan
df['total_pendapatan'] = df['harga'] * df['terjual']

#Groupby
ringkasan = df.groupby('menu')['total_pendapatan'].sum()
ringkasan = ringkasan.sort_values(ascending=False)

#output
print("\n==MENU LARIS==")
print(laris)

print("\n===PENJUALAN TERTINGGI===")
print(urut.head())

print("\n===TOTAL PENDAPATAN PER MENU===")
print(ringkasan)
```
